 B)   Handling Unexpected Values in Weight (Kilograms) and Freight Cost (USD) columns.

````python
# 1. Weight (Kilograms)

# View unexpected values in 'Weight (Kilograms)' column
df.loc[
    ~df['Weight (Kilograms)'].str.replace('.', '', regex=False).str.isnumeric(),
    'Weight (Kilograms)'
].unique()
````

````python
# Count unexpected values in Weight (Kilograms)' column
bad_count = pd.to_numeric(df['Weight (Kilograms)'], errors='coerce').isna().sum()
print(bad_count)
````

````python
# 2. Freight Cost (USD)

# View unexpected values in 'Freight Cost (USD)' column
df.loc[
    ~df['Freight Cost (USD)'].str.replace('.', '', regex=False).str.isnumeric(),
    'Freight Cost (USD)'
].unique()
````

````python
# Count unexpected values in 'Freight Cost (USD)' column
bad_count = pd.to_numeric(df['Freight Cost (USD)'], errors='coerce').isna().sum()
print(bad_count)
````

````python
'''
Unexpected values such as 'See DN-4234 (ID#:88632)' or 'See ASN-93 (ID#:1281)' indicate that the Weight/Freight
value should be retrieved from the record with the specified ID.
'''

# Step 1: Extract the referenced ID from the text
df['Ref_ID'] = (
    df['Weight (Kilograms)']
      .astype(str)
      .str.extract(r'ID#:(\d+)')
)

df['Ref_ID'] = pd.to_numeric(df['Ref_ID'], errors='coerce')


# Step 2: Convert the original columns to numeric
df['Weight_clean'] = pd.to_numeric(df['Weight (Kilograms)'], errors='coerce')

df['Freight_clean'] = pd.to_numeric(df['Freight Cost (USD)'], errors='coerce')


# Step 3: Create lookup dictionaries using the unique ID
id_to_weight = df.set_index('ID')['Weight_clean'].to_dict()

id_to_freight = df.set_index('ID')['Freight_clean'].to_dict()


# Step 4: Fill missing Weight values using the referenced ID
df['Weight_filled'] = df['Weight_clean']

mask = df['Weight_filled'].isna() & df['Ref_ID'].notna()

df.loc[mask, 'Weight_filled'] = (
    df.loc[mask, 'Ref_ID']
      .map(id_to_weight)
)


# Step 5: Fill missing Freight values using the referenced ID
df['Freight_filled'] = df['Freight_clean']

mask = df['Freight_filled'].isna() & df['Ref_ID'].notna()

df.loc[mask, 'Freight_filled'] = (
    df.loc[mask, 'Ref_ID']
      .map(id_to_freight)
)
````

````python
# Select all numerical columns to check for 0 and negative numbers

# Function to select only numeric columns excluding datetime64
def check_zero_negative(df, cols):
    result = {}

    for col in cols:
        result[col] = {
            'zero_count': (df[col] == 0).sum(),
            'negative_count': (df[col] < 0).sum()
        }

    return pd.DataFrame(result).T  # return result as a table
````

E) Variable Engineering

````python
# Date-related performance metrics

# 1: Time taken for the client to approve PQ and for FedEx to raise PO
df['PQ_to_PO_Days'] = (df['PO Sent to Vendor Date'] - df['PQ First Sent to Client Date']).dt.days

# 2: Vendor to Client delivery duration
df['PO_to_Delivery_Days'] = (df['Delivered to Client Date'] - df['PO Sent to Vendor Date']).dt.days

# 3: Actual delivery vs scheduled plan
df['Delivery_Delay_Days'] = (df['Delivered to Client Date'] - df['Scheduled Delivery Date']).dt.days

# 4: How long after physical delivery the system updates the record
df['Delivery_Recorded_Lag'] = (df['Delivery Recorded Date'] - df['Delivered to Client Date']).dt.days
````

F) Checking for Outliers

````python
# Select only the numerical variables where identifying outliers is meaningful
outlier_cols = [
    'Line Item Value',
    'Unit Price',
    'Pack Price',
    'Weight_filled',
    'Freight_filled',
    'Line Item Insurance (USD)',
    'Freight_Cost_Percentage',
    'Insurance_Percentage',
    'Cost_per_Kg',
    'PQ_to_PO_Days',
    'PO_to_Delivery_Days',
    'Delivery_Delay_Days',
    'Delivery_Recorded_Lag'
]

# Create an empty dictionary to store the outlier count for each variable
outlier_info = {}

# Calculate the number of outliers for each selected variable using the IQR method
for col in outlier_cols:

    # Calculate the first and third quartiles
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    # Calculate the Interquartile Range (IQR)
    IQR = Q3 - Q1

    # Define the lower and upper limits for identifying outliers
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Select rows where values fall outside the defined limits
    outliers = df[(df[col] < lower) | (df[col] > upper)]

    # Store the number of outliers for the current variable
    outlier_info[col] = outliers.shape[0]

# Convert the outlier counts into a DataFrame for easy comparison
outlier_summary = pd.DataFrame.from_dict(
    outlier_info,
    orient='index',
    columns=['Outlier Count']
)

# Display the variables along with their corresponding outlier counts
print(outlier_summary)
````

## **1. Column Standardization**

## **2. Handling Unexpected Values in Mixed-Type Columns**
Unexpected non-numeric values were found in **Weight (Kilograms)** and **Freight Cost (USD)**.  
Some cells contained reference strings such as `"See DN-4234 (ID# 88632)"`.

Steps taken:

- Extracted DN and ID references using regex.
- Cleaned DN values using a standardized numeric extraction.
- Created mapping tables using DN and ID to recover true weight and freight values.
- Filled missing numeric values using mapped references in a two-step process:
  1. Fill using Ref_ID  
  2. Fill using Ref_DN

---

## **3. Data Type Conversions**
- Date columns containing placeholders like `"Pre-PQ Process"` and `"Date Not Captured"` were replaced with `NaN` and converted to `datetime`.
- Converted **Line Item Insurance (USD)** from `object` to `float64` to enable numerical operations.

---

## **4. Numerical Data Cleaning and Validation**
- Zero and negative values were detected across cost, weight, and price columns.
- These were flagged using:
  - `is_zero_price`
  - `is_negative_price`
- One incorrect zero weight was imputed using the median weight per pack for the same **Brand** and **Item Description**.

---

## **5. Feature Engineering**
New variables were engineered to support the analysis:

### **Delivery Timeline Metrics**
- `PQ_to_PO_Days`
- `PO_to_Delivery_Days`
- `Delivery_Delay_Days`
- `Delivery_Recorded_Lag`

### **Delivery Categorization Metric**
To improve interpretability of delivery performance, a new categorical variable — `Delivery_Status` — was created using `Delivery_Delay_Days`.

This variable classifies each shipment into operationally meaningful groups:

- `Early`→ Delivered before the scheduled date  
- `On-Time`→ Delivered exactly on the scheduled date  
- `Delayed` → Delivered after the scheduled date  
- `Cancelled`→ No delivery recorded (missing `Delivered to Client Date`)

This transformation converts raw date differences into a business-friendly performance indicator, enabling deeper insights into vendor efficiency, shipment reliability, and operational delays.

### **Cost Efficiency Metrics**
- `Cost_per_Kg = Freight_filled / Weight_filled`
- `Insurance_percentage`
- `Freight_Cost_Percentage`
---

- Negative time durations were flagged as invalid.
- Identified inconsistencies indicating possible workflow or recording errors.

---

## **7. Outlier Detection**
The IQR method was applied to all numerical variables.

- Outliers were counted per variable.

  - Skewed shipping costs

````python
# Weight vs Line Item Insurance


# --------------------------------
# Prepare data
# --------------------------------

# Remove rows with missing weight or insurance values
insurance_weight = df.dropna(
    subset=['Weight_filled', 'Line Item Insurance (USD)']
).copy()

# Define weight ranges for analyzing median insurance cost
weight_bins = [0, 100, 500, 1000, 5000, 10000, 50000, 100000, float('inf')]

# Define labels for the weight ranges
weight_labels = [
    '0–100',
    '101–500',
    '501–1K',
    '1K–5K',
    '5K–10K',
    '10K–50K',
    '50K–100K',
    '100K+'
]

# Assign each shipment to a weight group
insurance_weight['Weight_Group'] = pd.cut(
    insurance_weight['Weight_filled'],
    bins=weight_bins,
    labels=weight_labels,
    include_lowest=True
)

# Calculate median insurance cost for each weight group
weight_insurance_median = (
    insurance_weight
    .groupby('Weight_Group', observed=True)['Line Item Insurance (USD)']
    .median()
)


# --------------------------------
# Create subplots
# --------------------------------

# Create two side-by-side plots for individual observations and median trends
fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=[
        'Weight vs Line Item Insurance',
        'Median Insurance Cost by Weight Group'
    ],
    horizontal_spacing=0.12
)


# --------------------------------
# 1. Individual observations
# --------------------------------

# Plot individual shipments to examine the relationship between weight and insurance cost
fig.add_trace(
    go.Scatter(
        x=insurance_weight['Weight_filled'],
        y=insurance_weight['Line Item Insurance (USD)'],
        mode='markers',
        name='Individual Shipments',
        opacity=0.5,
        hovertemplate=(
            '<b>Weight:</b> %{x:,.0f} kg<br>'
            '<b>Insurance:</b> $%{y:,.2f}'
            '<extra></extra>'
        )
    ),
    row=1,
    col=1
)


# --------------------------------
# 2. Median trend
# --------------------------------

# Plot the median insurance cost across different weight groups
fig.add_trace(
    go.Scatter(
        x=weight_insurance_median.index,
        y=weight_insurance_median.values,
        mode='lines+markers',
        name='Median Insurance',
        hovertemplate=(
            '<b>Weight Group:</b> %{x}<br>'
            '<b>Median Insurance:</b> $%{y:,.2f}'
            '<extra></extra>'
        )
    ),
    row=1,
    col=2
)


# --------------------------------
# Format axes and layout
# --------------------------------

# Label the x- and y-axes of the individual observations plot
fig.update_xaxes(
    title_text='Weight (Kg)',
    row=1,
    col=1
)

fig.update_yaxes(
    title_text='Line Item Insurance (USD)',
    row=1,
    col=1
)

# Label the x- and y-axes of the median trend plot
fig.update_xaxes(
    title_text='Weight Group (Kg)',
    row=1,
    col=2
)

fig.update_yaxes(
    title_text='Median Line Item Insurance (USD)',
    row=1,
    col=2
)

# Set the overall title, figure dimensions, and hide the legend
fig.update_layout(
    title='Relationship Between Shipment Weight and Line Item Insurance',
    width=1000,
    height=500,
    showlegend=False
)

# Display the interactive plots
fig.show()
````
